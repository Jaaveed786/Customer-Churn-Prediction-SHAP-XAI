# 01. Exploratory Data Analysis (EDA)

**Objective:** Understand the Telco Customer Churn dataset distributions, missing values, class imbalance, and key drivers of churn.

> **EDA Data Leakage Note:** EDA is intentionally conducted on the full dataset to derive general dataset insights. Preprocessing fits (scaling, imputation, one-hot encoding) are strictly fit ONLY on `X_train` in the modeling pipeline.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

DATA_PATH = Path('../data/raw/Telco-Customer-Churn.csv')
df = pd.read_csv(DATA_PATH)
print(f"Dataset shape: {df.shape}")
df.head()
```

```python
# Data Cleaning & Types
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
print(f"Missing TotalCharges: {df['TotalCharges'].isnull().sum()}")

df['Churn_Binary'] = (df['Churn'].str.strip().str.lower() == 'yes').astype(int)
print(f"Overall Churn Rate: {df['Churn_Binary'].mean():.2%}")
```

```python
# 1. Class Distribution
fig, ax = plt.subplots(figsize=(6, 4))
sns.countplot(data=df, x='Churn', palette=['#10b981', '#f43f5e'], ax=ax)
ax.set_title('Target Class Distribution (Churn vs Retained)')
plt.show()
```

```python
# 2. Tenure Distribution by Churn
fig, ax = plt.subplots(figsize=(8, 4))
sns.histplot(data=df, x='tenure', hue='Churn', kde=True, bins=30, palette=['#10b981', '#f43f5e'], ax=ax)
ax.set_title('Tenure Distribution by Churn Status')
plt.show()
```

```python
# 3. Contract Type vs Churn
contract_churn = df.groupby('Contract')['Churn_Binary'].mean().reset_index()
fig, ax = plt.subplots(figsize=(7, 4))
sns.barplot(data=contract_churn, x='Contract', y='Churn_Binary', palette='viridis', ax=ax)
ax.set_title('Churn Rate by Contract Type')
ax.set_ylabel('Churn Rate')
plt.show()
```
